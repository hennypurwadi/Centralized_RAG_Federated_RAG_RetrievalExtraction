## Extracting Data from Several Web Pages

```python
# Define URLs
url1 = "https://online.hull.ac.uk/courses/msc-artificial-intelligence"
url2 = "https://online.hull.ac.uk/courses/msc-artificial-intelligence#what-you-learn"
url3 = "https://online.hull.ac.uk/courses/msc-artificial-intelligence#course-modules"
url4 = "https://online.hull.ac.uk/courses/msc-artificial-intelligence#how-assessed"
url5 = "https://online.hull.ac.uk/courses/msc-artificial-intelligence#entry-requirements"
url6 = "https://online.hull.ac.uk/courses/msc-artificial-intelligence#career-prospects"
url7 = "https://online.hull.ac.uk/how-to-pay"
url8 = "https://online.hull.ac.uk/why-join-us/faqs"
url9 = "https://online.hull.ac.uk/funding-options"
url10 ="https://online.hull.ac.uk/course-costs" 

# List to store responses
responses = []

# Loop through each URL and fetch the content
for url in [url1, url2, url3, url4, url5, url6, url7, url8, url9, url10]:
    response = requests.get(url)
    if response.status_code == 200:  # Check if the request was successful
        soup = BeautifulSoup(response.text, 'html.parser')
        responses.append(soup)
    else:
        print(f"Failed to retrieve {url}")

# Collecting the text
text_data = []

for soup in responses:
    headings = soup.find_all(['h1', 'h2', 'h3'])
    paragraphs = soup.find_all('p')

    for heading in headings:
        if heading.text.strip():  # Check for non-empty content
            text_data.append(heading.text.strip())

    for paragraph in paragraphs:
        if paragraph.text.strip():  # Check for non-empty content
            text_data.append(paragraph.text.strip())

# Save the text data to a file with UTF-8 encoding
file_path = './data/msc_ai_hullonline_raw.txt'
with open(file_path, 'w', encoding='utf-8') as file:
    for line in text_data:
        file.write(line + "\n")

print(f"The text data has been saved to {file_path}")
```

## Summarizing Text Data

```python
# Load the text data from the file
file_path = './data/msc_ai_hullonline_raw.txt'
with open(file_path, 'r', encoding='utf-8') as file:
    text = file.read()
```

```python
sentences = sent_tokenize(text) # Tokenize the text into sentences
words = word_tokenize(text.lower()) # Tokenize the text into words
word_freq = Counter(words) # Calculate word frequencies

# Score each sentence based on word frequencies
sentence_scores = {}
for sentence in sentences:
    sentence_words = word_tokenize(sentence.lower())
    sentence_score = sum(word_freq.get(word, 0) for word in sentence_words)
    sentence_scores[sentence] = sentence_score

# Sort sentences by their scores and select the top ones
ranked_sentences = sorted(sentence_scores, key=sentence_scores.get, reverse=True)
num_sentences = max(int(len(sentences) * 0.3), 1)  # Adjust ratio to control summary length
summary = ' '.join(ranked_sentences[:num_sentences])

# Save the summary to a file
summary_file_path = './data/msc_ai_hullonline.txt'
with open(summary_file_path, 'w', encoding='utf-8') as file:
    file.write(summary)

print(f"The summary has been saved to {summary_file_path}")
```
