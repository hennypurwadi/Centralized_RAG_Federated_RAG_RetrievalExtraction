```python
from transformers import pipeline
```

## Load the summary from the file and print it

```python
# Load the summary from the file
def load_summary(file_path):
    with open(file_path, 'r', encoding='utf-8') as file:
        return file.read()

# Main execution
summary_file_path = './data/msc_ai_hullonline.txt'
loaded_summary = load_summary(summary_file_path)

# Print the loaded summary
print(loaded_summary)
```

## Define Questions

```python
questions = [
    'Who is Prof Adil Khan?',
    'What is the duration of the MSc AI program at the University of Hull?',
    'How much does the MSc AI course cost at the University of Hull?',
]
```

## Define Function for Question Answering Without Embeddings

```python
# Load the summary from the file
def load_summary(file_path):
    with open(file_path, 'r', encoding='utf-8') as file:
        return file.read()

# Function to ask & answer a single question
def ask_question(question, text, qa_pipeline):
    result = qa_pipeline(question=question, context=text)
    answer = result["answer"]
    print(f"Q: {question}\nA: {answer}\n")

# Function to ask a list of questions
def ask_questions(questions, text, qa_pipeline):
    for question in questions:
        ask_question(question, text, qa_pipeline)

# Function to perform QA with a chosen model
def perform_qa(model_name, questions, context_text):
    # Create the QA pipeline
    qa_pipeline = pipeline("question-answering", model=model_name)

    # Ask the list of questions
    ask_questions(questions, context_text, qa_pipeline)
```

### Perform Question Answering Without Embeddings with 'bert-tiny-5-finetuned-squadv2' model

```python
perform_qa("mrm8488/bert-tiny-5-finetuned-squadv2", questions, loaded_summary)
```

### Perform Question Answering Without Embeddings with 'distilbert-base-cased-distilled-squad' model

```python
perform_qa("distilbert-base-cased-distilled-squad", questions, loaded_summary)
```
